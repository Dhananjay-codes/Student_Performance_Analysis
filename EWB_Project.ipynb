```python
t_result=[]

for i in range(len(df['student_id'])):
    sum=df['maths'][i]+df['physics'][i]+df['chemistry'][i]+df['p_language'][i]+df['DSA'][i]
    t_result.append(sum)
    
    
df.insert(8,'total_theory_marks',t_result)
```

```python
l_result=[]

for i in range(len(df['student_id'])):
    sum=df['p_lang_lab'][i]+df['DSA_lab'][i]
    l_result.append(sum)

    
df.insert(11,'total_lab_marks',l_result)    
```

```python
df.head(3)
```

```python
overall_marks=[]


for i in range(len(df['student_id'])):
    sum=df['total_theory_marks'][i]+df['total_lab_marks'][i]+df['sports_score'][i]
    overall_marks.append(sum)

    
df.insert(14,'overall_marks',overall_marks)  
```

```python
df.head(3)
```

```python
i=0
grade=[]

while i<len(df['student_id']):
    if df['overall_marks'][i]>=770:
        grade.append('A++')
    elif (700<=df['overall_marks'][i]<770):
        grade.append('A+')
    elif(600<=df['overall_marks'][i]<700):
         grade.append('A')
    elif(550<=df['overall_marks'][i]<600):
          grade.append('B+')
    elif(500<=df['overall_marks'][i]<550):
          grade.append('B')
    elif(400<=df['overall_marks'][i]<500):
          grade.append('C')
    elif(350<=df['overall_marks'][i]<400):
          grade.append('D')
    else:
          grade.append('FAIL')
    i+=1


df.insert(15,'Grade',grade)
```

```python
position=[]

for i in range(len(df['student_id'])):
    if df['overall_marks'][i]>=350:
        position.append('PASS')
    else:
        position.append('FAIL')

df.insert(15,'position',position)
```

```python
i=0
toppers=[]

while i<len(df['student_id']):
    if df['overall_marks'][i]>=600:
        info=[df['name'][i],df['department'][i],df['overall_marks'][i]]
        toppers.append(info)
    i+=1    

    toppers_df=pd.DataFrame(
        toppers,
       columns=['Name', 'Department', 'Overall Marks']
    )

print('---------------TOPPERS-------------------------')
print()
print(toppers_df)        
```

```python
from IPython.display import display, Markdown

display(Markdown("### RANKING"))

ranking = df.dropna(subset=['overall_marks']).sort_values(
    by='overall_marks',
    ascending=False
).copy()

ranking['Rank'] = ranking['overall_marks'].rank(
    method='dense',
    ascending=False
).astype(int)

display(
    ranking[['student_id','Rank', 'name', 'department', 'overall_marks']].head(10)
)
```

```python
subjects = ['maths', 'physics', 'chemistry', 'p_language', 'DSA']

pass_fail = {}

for subject in subjects:
    passed = (df[subject] >= 35).sum()
    failed = (df[subject].notna() & (df[subject] < 35)).sum()
    pass_fail[subject] = [passed, failed]

result_df = pd.DataFrame(
    pass_fail,
    index=['Pass', 'Fail']
)

display(result_df)
```

```python
# Summary 


summary = pd.DataFrame({
    'Metric': [
        'Total Students',
        'Average Overall Marks',
        'Highest Overall Marks',
        'Lowest Overall Marks',
        'Average Attendance',
        'Missing Overall Marks'
    ],
    'Value': [
        len(df),
        round(df['overall_marks'].mean(), 2),
        df['overall_marks'].max(),
        df['overall_marks'].min(),
        round(df['attendance'].mean(), 2),
        df['overall_marks'].isnull().sum()
    ]
})

display(summary)
```

```python
# Department-wise Student Distribution


plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x='department',
    hue='department',
    palette='Set2',
    legend=False
)

for bar in ax.patches:
    ax.annotate(
        str(int(bar.get_height())),
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha='center',
        va='bottom'
    )

plt.title('Department-wise Student Distribution',
          fontsize=14, fontweight='bold')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.3)
sns.despine()
plt.tight_layout()
plt.show()
```

```python

attendance_category = []

for attendance in df['attendance']:
    if attendance <= 60:
        attendance_category.append('Low')
    elif attendance <= 75:
        attendance_category.append('Medium')
    else:
        attendance_category.append('High')

df['attendance_category'] = attendance_category

display(df[['name', 'attendance', 'attendance_category']].head(10))
```
